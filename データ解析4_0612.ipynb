```python
import pandas as pd

df = pd.DataFrame({
    "大人": [1200, 950, 420, 1500, 1650],
    "子供": [450, 320, 110, 850, 980],
    "天気": ["晴れ", "曇り", "雨", "晴れ", "晴れ"],
    "総客数": [1950, 1550, 620, 2750, 3050]
})

df
```

```python
X = pd.get_dummies(df[["大人", "子供", "天気"]], drop_first=True)
y = df["総客数"]

X
```

元データが

晴れ

曇り

雨

なら、変換後のイメージ。

   天気_曇り  天気_雨
0       0      0   # 晴れ
1       1      0   # 曇り
2       0      1   # 雨

ここで 晴れ が消えているのは、drop_first=True を付けているから。

なんで1つ落とすのか
理由は、全部の列を残すと情報がかぶるから です。

たとえば 晴れ, 曇り, 雨 の3列を全部残すと、

3列のうち2列が分かれば、残り1列は自動的に分かる
という状態になります。
これは重回帰では「同じ情報を二重に持っている」ことになって、計算が不安定になりやすいです。

なので、1つ基準を落とします。
今回なら 晴れ を基準にしていて、

天気_曇り = 1 なら曇り
天気_雨 = 1 なら雨
どちらも 0 なら晴れ
という読み方になります。

つまり、drop_first=True で落ちたのは 晴れ です。

「どこから曇りと雨は来たの？」
これは df["天気"] の中に もともと入っている値 です。
get_dummies() は、その列に登場した文字を自動で見つけて、列に展開します。

たとえばこのデータなら、

df["天気"]
の中身が

晴れ
曇り
雨
なので、get_dummies() がそれを見て自動で列を作ります。

つまり、曇りと雨はどこかから持ってきたのではなく、元の df に入っていた値から自動生成された ものです。

Pythonの get_dummies() は、その変換を自動でやってくれる道具です。
だから「天気の列を見て、必要なダミー列を勝手に作る」という理解で大丈夫です。

まずは覚えるポイント
get_dummies() = 文字を0/1列に変える
drop_first=True = 1列落として重複情報を避ける
落ちる列は、先頭のカテゴリ
今回は 晴れ が基準になった
曇り と 雨 は、元の df["天気"] から自動で作られた

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)

print("係数:", model.coef_)
print("切片:", model.intercept_)
```

```python
pred = model.predict(X)
df["予測値"] = pred
df["誤差"] = df["総客数"] - df["予測値"]
df
```

```python
df["誤差"].abs().mean()
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.DataFrame({
    "大人": [1200, 950, 420, 1500, 1650],
    "子供": [450, 320, 110, 850, 980],
    "天気": ["晴れ", "曇り", "雨", "晴れ", "晴れ"],
    "総客数": [1950, 1550, 620, 2750, 3050]
})

X = pd.get_dummies(df[["大人", "子供", "天気"]], drop_first=True)
y = df["総客数"]

X
```

```python
model = LinearRegression()
model.fit(X, y)

print("係数:", model.coef_)
print("切片:", model.intercept_)
```

```python
pred = model.predict(X)

df["予測値"] = pred
df["誤差"] = df["総客数"] - df["予測値"]
df
```

```python
df["誤差"].abs().mean()
```

つまり今回のデータでは、

予測値と実測値がほぼ一致している
誤差は実質ゼロ
ということです。

この e-13 は、コンピュータの小数計算の誤差みたいなものです。
人間の感覚では 0 と見て問題ありません。

```python
import pandas as pd

df = pd.read_csv("student-mat.csv", sep=";")
df.head()
```

```python
df.columns
```

ここは、どの列を説明変数に使えそうか。
重回帰なら、最初はこんな感じで考える。

studytime

failures

absences

G1

G2

目的変数は G3 にすることが多い。

これは「最終成績を、途中の成績や学習量から予測する」形\。

まずは数値列だけを使って、シンプルに。

```python
X = df[["studytime", "failures", "absences", "G1", "G2"]]
y = df["G3"]

X.head(), y.head()
```

(X.head(), y.head())
みたいに書くと、Pythonは2つのものをセットで返すので、見た目がこうなります。

(

  Xの先頭5行,

  yの先頭5件

)

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)

print("係数:", model.coef_)
print("切片:", model.intercept_)
```

```python
X.columns
```

```python
pred = model.predict(X)
df["予測G3"] = pred
df["誤差"] = df["G3"] - df["予測G3"]
df[["G3", "予測G3", "誤差"]].head()
```

## pred = model.predict(X)

これは、学習済みモデルに X を入れて、予測値を出している行です。

pred には、G3 の予測結果が入ります。

次の行で、予測値を df に追加しています。

## df["予測G3"] = pred

つまり、もともとの表に 予測G3 という新しい列を足しています。

その次は誤差です。

## df["誤差"] = df["G3"] - df["予測G3"]

これは
実測値 G3
予測値 予測G3
の差を取っているだけです。
最後に、見やすく抜き出しています。

## df[["G3", "予測G3", "誤差"]].head()

これは
実測値
予測値
誤差
だけを見ている、ということです。

なんで式を自分で作らずに計算できるのか
ここがいちばん本質で、モデルがデータから式を学習した結果。たとえば重回帰は、内部でこういう形の式を作っています。

G3 = a1 * studytime + a2 * failures + a3 * absences + a4 * G1 + a5 * G2 + b

この

a1, a2, a3, a4, a5

b

を、fit(X, y) がデータから自動で決めています。
だから、自分で式を毎回書かなくても、

# model.predict(X)

とするだけで、モデルが内部で持っている式を使って計算してくれます。
