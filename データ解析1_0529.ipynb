じゃあこれ逆に絞って3行3列だけ出力とかもできるの？

結論から言うと、完全にできる。 しかも、Pandasにはそのための「行と列を同時に指定して切り出す超強力な方法」が用意されている。C言語のようにループを回してインデックスを指定する必要はなく、指定の仕方も直感的である。

いくつか方法はあるが、実務で最も使われており、これだけ覚えておけば間違いないというloc（ロック）とiloc（アイロック）という方法を紹介する。

Google Colabの新しいセルで、以下のコードをそれぞれ試してみてほしい。

方法①：インデックスの「番号」で 3行 × 3列 を切り出す（iloc）
C言語の2次元配列の添字（matrix[i][j]）に一番近い感覚で、「0番目から数えて何行目・何列目か」の数字で指定する方法である。

Python
# 0〜2行目（3行分） と 0〜2列目（3列分）を切り出す
df.iloc[0:3, 0:3]
コードの解説
.iloc は「Integer Location（整数の位置）」の略である。

カンマ , を挟んで、左側が行の指定、右側が列の指定 になる。

0:3 はPythonのスライスという機能で、「0番目から、3番目の手前まで（つまり0, 1, 2）」 という意味になる。C言語の for (int i = 0; i < 3; i++) と同じ範囲の指定方法である。

方法②：列の「名前」で指定して切り出す（loc）
データサイエンスの実務では、「左から〇番目の列」と数字で数えるよりも、「『天気』と『大人』の列が欲しい」というように名前で指定することの方が圧倒的に多い。その場合は .loc を使う。

Python
# 0〜2行目（※3行分） と 「日付」「天気」「大人」の3列を切り出す
df.loc[0:2, ["日付", "天気", "大人"]]
コードの解説
.loc は「Location（位置・ラベル）」の略である。

列の指定部分（カンマの右側）に、["列名1", "列名2", "列名3"] のようにリスト形式で欲しい列の名前を直接書くことができる。

注意（C言語脳での落とし穴）： .loc で行を 0:2 と書いた場合、先ほどの .iloc と違って「2の手前まで」ではなく、「2番目の行も含める」 というルールになる。そのため、0, 1, 2の3行分が出力される。


```python
# 0〜2行目（3行分） と 0〜2列目（3列分）を切り出す
df.iloc[0:3, 0:3]
```

```python
# 0〜2行目（※3行分） と 「日付」「天気」「大人」の3列を切り出す
df.loc[0:2, ["日付", "天気", "大人"]]#.loc はLocation位置・ラベルの略で
```
